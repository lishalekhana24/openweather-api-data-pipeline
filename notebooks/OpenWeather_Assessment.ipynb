```python
from getpass import getpass

api_key = getpass("Enter your API key: ")
```

```python
import requests
from datetime import datetime

city = "Hyderabad"

url = "https://api.openweathermap.org/data/2.5/weather"

params = {
    "q": city,
    "appid": api_key
}

response = requests.get(url, params=params)

if response.status_code == 200:

    weather_data = response.json()

    print("Weather data collected successfully!")

    print("City:", weather_data["name"])
    print("Weather:", weather_data["weather"][0]["description"])
    print("Temperature:", weather_data["main"]["temp"])
    print("Humidity:", weather_data["main"]["humidity"])

else:
    print("Error:", response.status_code)
    print(response.text)
```

```python
import json
from datetime import datetime

# Create a timestamp
time_now = datetime.now().strftime("%Y%m%d_%H%M%S")

# Create file name
file_name = "raw_weather_" + time_now + ".json"

# Save weather data
with open(file_name, "w") as file:
    json.dump(weather_data, file, indent=4)

print("Raw weather data saved successfully!")
print("File name:", file_name)
```

```python
original_description = weather_data["weather"][0]["description"]

print("Original Weather:", original_description)
```

```python
normalized_description = original_description.title()

print("Normalized Weather:", normalized_description)
```

```python
if "clear" in original_description.lower():

    category = "Clear"

elif "cloud" in original_description.lower():

    category = "Cloudy"

elif "rain" in original_description.lower() or "drizzle" in original_description.lower():

    category = "Rainy"

else:

    category = "Other"


print("Weather Category:", category)
```

```python
city_name = weather_data["name"]

observation_time = datetime.fromtimestamp(weather_data["dt"])

data = {
    "City": [city_name],
    "Original Description": [original_description],
    "Normalized Description": [normalized_description],
    "Weather Category": [category],
    "Observation Timestamp": [observation_time]
}

df = pd.DataFrame(data)

df.to_csv("transformed_weather.csv", index=False)

print("Transformed CSV saved successfully!")

display(df)
```

```python
# Check required columns

if "City" in df.columns and "Original Description" in df.columns and "Normalized Description" in df.columns and "Weather Category" in df.columns and "Observation Timestamp" in df.columns:

    print("All required columns are present!")

else:

    print("Some columns are missing!")


# Check empty values

if df.isnull().sum().sum() == 0:

    print("No missing values found!")

else:

    print("Missing values found!")


# Check number of rows

print("Total records:", len(df))
```

```python
import smtplib
from email.message import EmailMessage
from getpass import getpass

print("Email libraries ready!")
```

```python
sender_email = input("Enter your Gmail address: ")

receiver_email = input("Enter receiver Gmail address: ")

app_password = getpass("Enter your Gmail App Password: ")

print("Email details received!")
```

```python
# Send email using Gmail SSL

if df.isnull().sum().sum() == 0 and len(df) > 0:

    try:
        message = EmailMessage()
        message["Subject"] = "OpenWeather Project - Success"
        message["From"] = sender_email
        message["To"] = receiver_email
        message.set_content(
            "Your Hyderabad weather pipeline completed successfully. "
            "The output CSV passed validation."
        )

        server = smtplib.SMTP_SSL("smtp.gmail.com", 465, timeout=30)
        server.login(sender_email, app_password.replace(" ", ""))
        server.send_message(message)
        server.quit()

        print("Success email sent!")

    except Exception as e:

        print("Email Error:", e)

else:

    print("Validation failed. Email not sent!")
```
